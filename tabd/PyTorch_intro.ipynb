# Простые операции с pytorch

`Тензор (tensor)` - это математический абстрактный объект, который обобщает понятия скаляра, вектора и матрицы.

**Основные характеристики тензоров:**

- `Ранг:` Ранг тензора определяет, сколько индексов ему требуется для полной спецификации. Скаляр (тензор нулевого ранга) не требует индексов, вектор (тензор первого ранга) требует один индекс, матрица (тензор второго ранга) - два индекса, и так далее.

- `Размерности:` Тензоры имеют размерности, которые определяют, сколько элементов содержится вдоль каждой из их осей.

- `Компоненты:` Элементы тензора называются его компонентами. Компоненты тензора могут быть числами, функциями или другими объектами в зависимости от контекста.

```python
import torch

# создаем тензор ранга 4 размером 2x2x2x2 со случайными значениями
tensor_torch = torch.ones(2, 2, 2, 2)

# Выводим созданный тензор
print(tensor_torch)

# Определяем объем памяти, занимаемый тензором в байтах
memory_bytes = tensor_torch.element_size() * tensor_torch.nelement()

print(f"Объем памяти, занимаемый тензором: {memory_bytes} байт")
```

```python
import numpy as np

# Создаем тензор ранга 4 размером 2x2x2x2 со случайными значениями
tensor_numpy = np.ones((2, 2, 2, 2))

# Выводим созданный тензор
print(tensor_numpy)


# Определяем объем памяти, занимаемый массивом в байтах
memory_bytes = tensor_numpy.nbytes

# Выводим объем памяти
print(f"Объем памяти, занимаемый массивом: {memory_bytes} байт")
```

**В чем же разница?**.

1) Автоматическое дифференцирование:
- `NumPy:` NumPy не предоставляет встроенной поддержки для автоматического дифференцирования. Это означает, что вы должны явным образом вычислять градиенты при необходимости, что может быть неудобно при разработке сложных моделей машинного обучения.

- `PyTorch:` PyTorch обеспечивает автоматическое дифференцирование через свой модуль `autograd`. Это позволяет легко вычислять градиенты функций и использовать их в процессе обучения нейронных сетей. Эта функциональность делает PyTorch более подходящим для глубокого обучения и исследований.

2) Оптимизация на GPU
 - `NumPy:` NumPy поддерживает вычисления на CPU, но для оптимизации на GPU требуется использовать дополнительные библиотеки, такие как CuPy.

 - `PyTorch:` PyTorch предоставляет встроенную поддержку для оптимизации на графических процессорах (GPU) с помощью CUDA. Это позволяет ускорять обучение нейронных сетей и другие вычисления, используя мощности GPU.

Плюс Numpy:  
`Производительность на CPU:`   
 В некоторых случаях NumPy может предоставить более оптимизированные вычисления на CPU, особенно если вы работаете с большими массивами данных и используете оптимизированные библиотеки BLAS (Basic Linear Algebra Subprograms).

# Про градиенты (простой пример на функции от одной переменной)

Если мы хотим вычислить градиенты, то у тензора должен быть  параметр requires_grad=True

```python
x = torch.tensor([[1., -1.], [1., 1.]], requires_grad=True)
x
```

```python
out = ((2*x+3)**2).sum()
out
```

$((2x+3)^2)' = 2*(2x+3)*(2x+3)' = 2*(2x+3)*2 = 4*(2x+3)$



вызов метода backward() на переменной out запускает процесс вычисления градиентов. Этот процесс начинается с переменной out и распространяется назад через граф вычислений.

```python
out.backward()
```

```python
out
```

```python
x.grad
```

# простая нейронка

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from matplotlib import pyplot as plt
```

```python
X = np.linspace(0, 1, 100)
alpha = 10.
beta = -3.
sigma = 0.5

Y = alpha*X + beta + sigma*np.random.randn(len(X))
```

```python
X
```

```python
plt.scatter(X,Y)
```

`nn.Module` - это базовый класс в библиотеке PyTorch, который используется для определения нейронных сетей и других моделей машинного обучения. Он предоставляет удобные методы для управления параметрами модели, а также для определения прямого и обратного проходов (forward и backward passes) через сеть.

`super().__init__()` - это вызов конструктора суперкласса в Python. В данном контексте `super()` возвращает объект, который представляет родительский класс `nn.Module`, а `__init__()` - это метод этого объекта, который выполняет инициализацию объекта суперкласса.

```python
class LinearModel(nn.Module):
    def __init__(self, alpha, beta):
        super().__init__()
        self.alpha = nn.Parameter(torch.tensor(alpha))
        self.beta = nn.Parameter(torch.tensor(beta))
    def forward(self, x):
        return self.alpha*x+self.beta
```

```python
model_lr =  LinearModel(0.1, 0.)
```

```python
res_out = model_lr(torch.from_numpy(X))
```

```python
res_out
```

```python
# res_out.detach()
res_out.data
```

```python
plt.plot(X,
         res_out.data,
         color = 'green',
         label='model')
plt.scatter(X,Y,  marker='o', s=4, label='data',)
plt.legend()
plt.show()
```

```python
model_lr.alpha
```

```python
model_lr.beta
```

```python
for param in model_lr.parameters():
  print(param)
```

```python
# подготовим данные для обучения модели
X_torch , Y_torch = torch.tensor(X.reshape([-1,1])), torch.tensor(Y.reshape([-1,1]))

```

```python
# инициализируем модель и её оптимизатор
# model_lr =  LinearModel(1., 2.)
optimizer = optim.Adam(model_lr.parameters())
```

```python
optimizer.state
```

```python
# обучим модель на 10 эпохах
for i in range(10):
    y_pred = model_lr(X_torch)
    e = ((y_pred-Y_torch)**2).mean()
    e.backward()
```

```python
model_lr.alpha, model_lr.beta, e
```

ничего не изменилось , но так и должно быть, мы же не добавили шаг optimizer

```python
optimizer.state
```

```python
# добавим "шаг" оптимизитора!
for i in range(10):
    y_pred = model_lr(X_torch)
    e = ((y_pred-Y_torch)**2).mean()
    e.backward()
    optimizer.step()
```

```python
model_lr.alpha, model_lr.beta, e
```

```python
# шаги накапливаются!
optimizer.state
```

```python
plt.plot(X,
         y_pred.data,
         color = 'green',
         label='model')
plt.scatter(X,Y,  marker='o', s=4, label='data',)
plt.legend()
plt.show()
```

```python
# добавим обнуление градиента (на каждом ли шаге?)
# optimizer.zero_grad()

for i in range(10):
    optimizer.zero_grad()
    y_pred = model_lr(X_torch)
    e = ((y_pred-Y_torch)**2).mean()
    e.backward()
    optimizer.step()
```

```python
model_lr.alpha, model_lr.beta, e
```

```python
optimizer.state
```

```python
# иногда надо , чтобы один параметр обучался, а второй нет
model_lr.alpha.requires_grad=False
```

```python
# добавим "шаг" оптимизитора!
for i in range(10):
    optimizer.zero_grad()
    y_pred = model_lr(X_torch)
    e = ((y_pred-Y_torch)**2).mean()
    e.backward()
    optimizer.step()
    print(model_lr.alpha, model_lr.beta)
```

```python
model_lr.alpha, model_lr.beta, e
```

```python
model_lr.alpha.grad # градиента нет
```

```python
# контроль значения градиента
from torch.nn.utils import clip_grad_norm_

model_lr =  LinearModel(0.5, 0.5)
optimizer = optim.Adam(model_lr.parameters(), lr=0.01)

for i in range(10):
    optimizer.zero_grad()
    y_pred = model_lr(X_torch)
    e = ((y_pred-Y_torch)**2).mean()
    e.backward()

    # clip_grad_norm_(model_lr.parameters(), 1, foreach=False)
    clip_grad_norm_(model_lr.alpha, 1, foreach=False)

    optimizer.step()

    # print('alpha=', model_lr.alpha, model_lr.alpha.grad)
    print('beta=', model_lr.beta, model_lr.beta.grad)
```

```python
# используем встроенную фунцию потерь

model_lr =  LinearModel(0.5, 0.5)
loss = nn.MSELoss()
optimizer = optim.Adam(model_lr.parameters(), lr=1)


for i in range(10):
    optimizer.zero_grad()
    y_pred = model_lr(X_torch)
    e = loss(y_pred, Y_torch)
    e.backward()

    clip_grad_norm_(model_lr.parameters(), 1)

    optimizer.step()

    print(model_lr.alpha, model_lr.alpha.grad)
```
