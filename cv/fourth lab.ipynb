```python
import cv2
import numpy as np

def custom_threshold(image, t):
    binary = (image >= t).astype(np.uint8) * 255
    return binary

def custom_erode_cross(binary_image, iterations):
    res = binary_image.copy()
    h, w = binary_image.shape
    
    for _ in range(iterations):
        temp_img = res.copy()
        for y in range(1, h - 1):
            for x in range(1, w - 1):
                # крестик
                center = temp_img[y, x]
                up     = temp_img[y-1, x]
                down   = temp_img[y+1, x]
                left   = temp_img[y, x-1]
                right  = temp_img[y, x+1]
                
                # Эрозия — это выбор минимального значения в окрестности 
                res[y, x] = min(center, up, down, left, right)
    return res

def on_trackbar(val):
    thresh_val = cv2.getTrackbarPos('Threshold', 'Morfology')
    iters = cv2.getTrackbarPos('Iterations', 'Morfology')
    
    # 1. Ручные функции
    manual_bin = custom_threshold(original_img, thresh_val)
    manual_eroded = custom_erode_cross(manual_bin, iters)
    
    _, cv_bin = cv2.threshold(original_img, thresh_val, 255, cv2.THRESH_BINARY)
    kernel = cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3)) 
    cv_eroded = cv2.erode(cv_bin, kernel, iterations=iters)
    
    combined = np.hstack((original_img, manual_eroded, cv_eroded))
    cv2.imshow('Morfology', combined)


filename = '1.jpg' 
original_img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

if original_img is None:
    print("Ошибка загрузки")
else:
    cv2.namedWindow('Morfology')
    cv2.createTrackbar('Threshold', 'Morfology', 127, 255, on_trackbar)
    cv2.createTrackbar('Iterations', 'Morfology', 1, 5, on_trackbar)
    on_trackbar(30)
    while True: 
        if cv2.waitKey(1) & 0xFF == 27: 
            break
    cv2.destroyAllWindows()
```
