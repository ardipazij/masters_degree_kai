```python
import cv2
import numpy as np

def custom_average_blur(image, kernel_size):
    height, width = image.shape
    result = np.zeros((height, width), dtype=np.uint8)
    
    offset = kernel_size // 2
    num_pixels = kernel_size * kernel_size
    
    img_float = image.astype(np.float32)

    # Проходим по пикселям, оставляя границы нетронутыми
    for y in range(offset, height - offset):
        for x in range(offset, width - offset):
            window = img_float[y - offset : y + offset + 1, x - offset : x + offset + 1]
            
            # Среднее арифметическое яркостей по окрестности
            result[y, x] = int(np.sum(window) / num_pixels)
            
    return result

def on_trackbar(val):
    size = cv2.getTrackbarPos('Kernel Size', 'Averaging Filter')
    if size < 1: size = 1
    if size % 2 == 0: size += 1 
    
    custom_blurred = custom_average_blur(original_img, size)
    
    cv_blurred = cv2.blur(original_img, (size, size))
    
    combined = np.hstack((original_img, custom_blurred, cv_blurred))
    cv2.imshow('Averaging Filter', combined)

filename = '1.jpg' 
original_img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

if original_img is None:
    print(f"Ошибка: Не удалось загрузить {filename}")
else:
    cv2.namedWindow('Averaging Filter')
    cv2.createTrackbar('Kernel Size', 'Averaging Filter', 3, 21, on_trackbar)
    on_trackbar(3)
    
    while True:
        if cv2.waitKey(1) & 0xFF == 27: break
    cv2.destroyAllWindows()
```
