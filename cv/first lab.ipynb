```python
import cv2
import numpy as np

# Функция, которая вызывается при движении ползунка
def update(val):
    # val - это значение с ползунка (от 0 до 255)
    
    # ФОРМУЛА: s = val - r
    # 1. Сначала переводим картинку в int (целые числа), 
    # чтобы можно было получать отрицательные значения при вычитании
    r = img.astype(int)
    
    # 2. Вычитаем пиксели из значения ползунка
    # Если ползунок на 255 -> классический негатив
    # Если ползунок меньше -> картинка темнеет
    res = val - r
    
    # 3. НОРМАЛИЗАЦИЯ (Clipping)
    # np.clip обрезает всё, что вышло за границы:
    # если число < 0, оно станет 0
    # если число > 255, оно станет 255
    res_clipped = np.clip(res, 0, 255)
    
    # 4. Превращаем обратно в формат картинки (uint8)
    res_final = res_clipped.astype(np.uint8)

    # Показываем
    cv2.imshow('Manual Negative', res_final)

# --- ОСНОВНАЯ ЧАСТЬ ---
img = cv2.imread('4.jpg')

if img is None:
    print("Картинка не найдена!")
else:
    # Создаем окно
    cv2.namedWindow('Manual Negative')

    # Создаем ползунок 'Invert Level'
    # Аргументы: Имя, Окно, Начальное значение, Максимум, Функция обновления
    cv2.createTrackbar('Level', 'Manual Negative', 255, 255, update)

    # Запускаем функцию один раз, чтобы картинка появилась сразу
    update(255)

    print("Нажми любую клавишу для выхода")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
