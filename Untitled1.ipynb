```python
import random
message = 'aaababaabadddccbbbbbaaabbbddccdddaaaccccaaaaaacabb' * 40
stages = sorted(set(message))
n = len(stages)
C = [[0 for _ in range(n)] for _ in range(n)]

for a, b in zip(message[:-1], message[1:]):
    i, j = state_to_idx[a], state_to_idx[b]
    C[i][j] += 1

print("\nМатрица переходов C (частоты):")
for row in C:
    print(row)

P = []
for row in C:
    s = sum(row)
    if s == 0:
        P.append([0 for _ in row])
    else:
        P.append([count / s for count in row])

print("\nМатрица переходов P (вероятности):")
for row in P:
    print([round(x, 3) for x in row])

generated = [message[0]]
for _ in range(len(message) - 1):
    current = generated[-1]            
    i = state_to_idx[current]              
    probs = P[i]                          
    r = random.random()
    cum_sum = 0.0
    for j, p in enumerate(probs):         
        cum_sum += p
        if r <= cum_sum:
            generated.append(idx_to_state[j])
            break

generated_str = "".join(generated)
print("\nСгенерированная строка:")
print(generated_str)

C_gen = [[0 for _ in range(n)] for _ in range(n)]
for a, b in zip(generated_str[:-1], generated_str[1:]):
    i, j = state_to_idx[a], state_to_idx[b]
    C_gen[i][j] += 1

P_gen = []
for row in C_gen:
    s = sum(row)
    if s == 0:
        P_gen.append([0 for _ in row])
    else:
        P_gen.append([count / s for count in row])
print("\nМатрица переходов сгенерированной строки (частоты):")
for row in C_gen:
    print(row)

print("\nМатрица переходов сгенерированной строки (вероятности):")
for row in P_gen:
    print([round(x, 3) for x in row])

matches = sum(a == b for a, b in zip(message, generated_str))
similarity = matches / len(message)
print(f"\nСовпадений: {matches}/{len(message)} = {similarity:.2%}")
```
